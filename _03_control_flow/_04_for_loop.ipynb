# for 반복문

- `for` 반복문은 iterable 객체의 요소를 하나씩 꺼내 실행한다.
- `str`, `list`, `tuple`, `dict`, `set`, `range` 등이 대표적인 iterable 객체이다.
- 반복 횟수가 정해져 있거나 컬렉션의 모든 요소를 순회할 때 자주 사용한다.

## 1. 문자열, 리스트, 튜플 순회

문자열은 문자 단위로, 리스트와 튜플은 요소 단위로 순회한다.

```python
from operator import index
from unittest import result

#for A in B
# A: 객체에서 꺼내와 저장할 변수, 요소 한 개
# B: 순회(반복) 가능한 객체(str, list, tuple, dict, set

text:str='hello world'
for ch in text:
    print(ch)

lst:list[int]=[1,2,3]
for num in lst:
    print(num)

tpl:tuple[int]=(10,20,30)
for num in tpl:
    print(num)
```

## 2. dict와 set 순회

`dict`를 그대로 순회하면 key가 나온다. `items()`를 사용하면 key와 value를 함께 꺼낼 수 있다.

`set`은 순서가 없는 자료형이므로 출력 순서는 실행 환경에 따라 달라질 수 있다.

```python
from typing import Any
user_info:dict[str,Any] ={
                #Any: typing 모듈에 있는 문법 / 모든타입을 의미(클래스)
    'name':'홍길동',
    'age':18,
    'hobby':['독서','유튜브']
}

#dict 순회 -> key순회, value순회, item순회(+ unpacking)
print('--- dict key 순회 ---')
for key in user_info:
    print(f'{key}: {user_info[key]}')

print('--- dict item 순회(unpacking) ---')
for key, value in user_info.items():
                #user_info.items() : item 단위로 조회됨
                #key, value : (key:value)를 unpacking해 각각을 조회
    print(f'{key}: {value}')

print('--- set 순회 : if문을 곁들인... ---')
nums:set[int]={1,2,3,3,3,9,6,6,1,33,333,303}

#결과 저장용 list(빈 list에 결과를 누적해 추가)
results:list[int]=[]

for num in nums:
    if num%2==0:
        results.append(num)
print(results)
```

## 3. range 객체

`range(start, end, step)`은 정수 범위를 표현하는 객체이다.

- `start`: 시작값
- `end`: 끝값, 포함되지 않음
- `step`: 증가 또는 감소 간격

```python
#range 객체: 지정된 범위 내 정수를 step만큼 증가하면서 값을 하나씩 반환하는 객체

print('--- range 객체 ---')
rng=range(1,5,1)        #-> 1,2,3,4
print(rng,type(rng))    #range(1, 5) <class 'range'>

for num in rng:
    print(num)

print('--- range 객체를 이용한 정수 list 만들기 ---')
nums:list[int]=list(range(1,11,1))
print('nums: ',nums)

print('--- 1부터 100까지 모든 정수의 합 구하기 ---')
n_sum=0             #연산을 누적시킬 변수 생성
for n in range(1,101,1):
    n_sum += n      #n_sum = n_sum + n
print('n_sum: ',n_sum)

print('--- 1부터 100까지 모든 홀수의 합 구하기 ---')
n_sum=0

for n in range(1,101,2):
    n_sum += n

#if문을 활용한 for문
#for n in range(1,101,1):
#    if n%2==1:
#        n_sum += n

print('n_sum: ',n_sum)
```

## 4. 중첩 반복문

반복문 안에 반복문을 넣으면 행과 열처럼 2차원 구조를 만들 수 있다. 구구단과 별찍기 같은 예제에 자주 사용한다.

```python
print('--- 중첩 반복문 예시 ---')
for a in range(1,4):        #a=1,2,3
    for b in range(1,3):    #b=1,2
        print(a,b)

print('--- 구구단 2단 출력하기 ---')
dan:int=2
for num in range(1,10):
    print(f'{dan}x{num}={num*dan}')

print('--- 구구단 전체 출력(2~9단) ---')
for dan in range(2,10):
    for num in range(1,10):
        print(f'{dan}x{num}={num*dan}')
    print("-" * 20)         #단마다 선으로 구분

print('--- range를 이용한 반복 횟수 설정 ---')
print('- ❤️ 3행 5열 출력 -')
for row in range(3):  #range(3)=0,1,2
    for col in range(5):
        print('❤️', end='')  #end 속성: print() 종료 후 출력할 문자 지정
    print()
```

## 5. zip과 enumerate

- `zip()`은 여러 iterable을 같은 순서끼리 묶는다.
- `enumerate()`는 인덱스와 요소를 함께 꺼낸다.

```python
print('--- zip 객체 ---')
names:list[str]=['a','b','c']
nums:list[int]=[10,20,30]

z=zip(names,nums)
print(z,type(z))

for name, num in z:
    print(name, num)        #같은 인덱스끼리 묶어서 출력

print('--- enumerate 객체 ---')
foods:list[str]=['엽떡','마라탕','닭찌찌']
for index, food in enumerate(foods):
    print(f'{index}번 인덱스: {food}')
```

## 6. 반복문으로 평균 구하기

반복문을 사용하면 리스트나 딕셔너리 리스트에 들어 있는 값을 누적해서 평균을 계산할 수 있다.

```python
print('--- list에 담긴 값의 평균 구하기 ---')
scores:list[int]=[20,43,91,49,60]
score_sum=0
for score in scores:
    score_sum += score              #랍계 누적
score_avg=score_sum/len(scores)     #합계 / 요소의 개수 = 평균
print('평균: ',score_avg)

#for문 없이 평균 구하는 법
# 숫자로만 구성된 list or tuple일 때만 사용 가능
print('평균: ', sum(scores)/len(scores))
```

```python
print('--- list[dict]에 담긴 값의 평균 구하기 ---')
students:list[dict[str,object]]=[
    {'name':'홍길동', 'score':100},
    {'name':'신사임당', 'score':77},
    {'name':'이순신', 'score':85}
]
#score 조회 확인
#print(students[2].get('score'))

# list 순회를 이용한 평균 계산
sum:int=0
for std in students:
    sum += std['score']     #std = students[0], students[1], students[2]

avg=sum/len(students)
print('평균: ',round(avg))

# range + indexing을 이용한 평균 계산
for index in range(len(students)):
            #range(len(students))=range(3)=0,1,2 -> index에 저장
    sum += students[index].get('score')

avg = sum / len(students)
print("평균: ", round(avg, 2) ) # round(): 반올림
```
